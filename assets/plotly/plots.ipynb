```python
import numpy as np
import plotly.graph_objects as go
from scipy.stats import norm

# Generate fixed samples from a normal distribution
np.random.seed(42)
samples = np.random.normal(loc=0, scale=1, size=1000)

# Define the range for the x-axis (for the PDF)
x = np.linspace(-5, 5, 500)

# Initial mean and variance
initial_mean = 0
initial_variance = 1

# Calculate the initial PDF
pdf = norm.pdf(x, loc=initial_mean, scale=np.sqrt(initial_variance))

# Create the histogram and initial PDF
fig = go.Figure()

# Histogram of samples
fig.add_trace(go.Histogram(
    x=samples,
    histnorm='probability density',
    name='Histogram',
    opacity=0.7
))

# Overlay PDF
pdf_trace = go.Scatter(
    x=x,
    y=pdf,
    mode='lines',
    name='Normal PDF',
    line=dict(color='red')
)
fig.add_trace(pdf_trace)

# Create update menus for sliders
mean_values = np.linspace(-2, 2, 9)  # Mean values for the slider
variance_values = np.linspace(0.5, 2.5, 9)  # Variance values for the slider

# Create sliders
sliders = [
    # Slider for mean
    {
        "active": 4,
        "currentvalue": {"prefix": "Mean: "},
        "steps": [
            {
                "args": [
                    {"y": [norm.pdf(x, loc=mean, scale=np.sqrt(initial_variance))]},
                    {"title": f"Mean: {mean}, Variance: {initial_variance:.2f}"}
                ],
                "label": f"{mean:.2f}"
            }
            for mean in mean_values
        ]
    },
    # Slider for variance
    {
        "active": 2,
        "currentvalue": {"prefix": "Variance: "},
        "steps": [
            {
                "args": [
                    {"y": [norm.pdf(x, loc=initial_mean, scale=np.sqrt(variance))]},
                    {"title": f"Mean: {initial_mean:.2f}, Variance: {variance:.2f}"}
                ],
                "label": f"{variance:.2f}"
            }
            for variance in variance_values
        ]
    }
]

# Add sliders to the layout
fig.update_layout(
    title="Statistical Inference by Eye",
    xaxis_title="Value",
    yaxis_title="Density",
    sliders=sliders
)

# Show the plot
fig.show()

# Save the plot as an HTML file
fig.write_html("stat_by_eye.html")
```
